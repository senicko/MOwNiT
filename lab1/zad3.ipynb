```python
def dzeta(s: float, n: int, reversed=False, precision=np.float32):
    s = precision(s)

    iterator = range(1, n) if not reversed else range(n, 0 , -1)
    result = precision(0)

    for k in iterator:
        result += 1 / (k ** s)

    return result

def eta(s: float, n: int, reversed=False, precision=np.float32):
    s = precision(s)

    iterator = range(1, n) if not reversed else range(n, 0 , -1)
    result = precision(0)

    for k in iterator:
        result += ((-1) ** k) * (1 / k ** s)

    return result
```

```python
S = [np.float32(2), np.float32(3.6667), np.float32(5), np.float32(7.2), np.float32(10)]
N = [50, 100, 200, 500, 1000]
```

```python
for s in S:
    for n in N:
        result_dzeta_forward_f32 = dzeta(s, n)
        result_dzeta_backward_f32 = dzeta(s, n, reversed=True)

        result_eta_forward_f32 = eta(s, n)
        result_eta_backward_f32 = eta(s, n, reversed=True)

        result_dzeta_forward_f64 = dzeta(s, n, precision=np.float64)
        result_dzeta_backward_f64 = dzeta(s, n, reversed=True, precision=np.float64)

        result_eta_forward_f64 = eta(s, n, precision=np.float64)
        result_eta_backward_f64 = eta(s, n, reversed=True, precision=np.float64)

        print(f"""
----
s={s} n={n}

eta:
forward_f32 \t{result_eta_forward_f32}\tforward_f64\t{result_eta_forward_f64}
backward_f32\t{result_eta_backward_f32}\tbackward_f64\t{result_eta_backward_f64}

dzeta:
forward \t{result_dzeta_forward_f32}\tforward_f64\t{result_dzeta_forward_f64}
backward\t{result_dzeta_backward_f32}\tbackward_f64\t{result_dzeta_backward_f64}
---
""")
```
