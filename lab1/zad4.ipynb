Chaos deterministyczny, np. efekt motyla

Odwzorowanie logistyczne jest przykładem chaosu deterministycznego. Takie układy są bardzo wrażliwe na parametry. Zaburzenie na poziomie błędu reprezentacji drastycznie zmienia trajektorię układu równań.

```python
import numpy as np
import time 
import matplotlib.pyplot as plt
```

```python
iters = 10 ** 5
results = []

for r in np.linspace(1, 4, 4):
    for x in np.linspace(0, 1, 10):
        result = np.float64(x)

        for _ in range(iters):
            result = r * result * (1 - result)

        results.append((x, results))
```

```python
plt.scatter(results)
```
