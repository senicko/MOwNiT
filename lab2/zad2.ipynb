# Zad 2 - Dekompozycja LU

Dekompozycja LU polega na faktoryzacji macierzy $A$ na iloczyn dwóch macierzy, jednej trójkątnej dolnej $L$, a drugiej trójkątnej górnej $U$.

$
A = LU
$

W celu przeprowadzenia faktoryzacji zapisujemy kolejne współczynniki wykorzystywane podczas eliminacji Gaussa - sprowadzającej macierz $U$ do postaci trójkątnej górnej.

```python
def lu(A: np.array) -> np.array:
    """
    Performs in-situ LU decomposition overwriting A contents.
    """

    n, _ = A.shape

    for i in range(n):
        for j in range(i + 1, n):
            factor = A[j, i] / A[i, i]

            A[j, i:] -= A[i, i:] * factor
            A[j, i] = factor

def lu_components(LU: np.array) -> tuple[np.array, np.array]:
    """
    Extracts L and U components from in-situ LU decomposition.
    """

    n, _ = LU.shape

    L = np.identity(n, dtype=np.float64)
    U = np.zeros(shape=(n, n), dtype=np.float64)

    for i in range(n):
        for j in range(n):
            if i > j:
                L[i, j] = LU[i, j]
            else:
                U[i, j] = LU[i, j]

    return L, U
```

```python
def print_readable_matrix(mat, precision=2):
    for i, row in enumerate(mat):
        for val in row:
            print(f"{val:17.{precision}f}", end="")
        print()
```

```python
n = 10
A = np.random.rand(n, n).astype(np.float64)
```

```python
A2 = A.copy()
lu(A2)
L, U = lu_components(A2)
result = A - L @ U
print_readable_matrix(np.abs(A - L @ U), precision=12)
print(f"{np.linalg.norm(result):.2f}")
```
