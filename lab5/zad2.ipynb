```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
```

```python
# Load the raccoon face image
image = scipy.datasets.face(gray=True)
print(image.shape)
```

```python
display(image)
```

```python
plt.imshow(image, cmap='gray')
plt.axis("off")
plt.show()
```

```python
def approx_image(image: np.ndarray, r: int) -> np.ndarray:
    u, s, vt = np.linalg.svd(image)
    return u[:, :r] @ np.diag(s[:r]) @ vt[:r, :]
```

```python
plt.imshow(approx_image(image, 1), cmap='gray')
plt.axis("off")
plt.show()
```

```python
plt.imshow(approx_image(image, 10), cmap='gray')
plt.axis("off")
plt.show()
```

```python
plt.imshow(approx_image(image, 20), cmap='gray')
plt.axis("off")
plt.show()
```

```python
plt.imshow(approx_image(image, 30), cmap='gray')
plt.axis("off")
plt.show()
```
